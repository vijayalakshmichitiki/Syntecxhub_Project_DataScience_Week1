```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load data
df = pd.read_csv("sales_data_.csv")
```

```python
df
```

```python
df.info()
```

```python
# Convert date column
df['date'] = pd.to_datetime(df['date'])
```

```python
# Add month column
df['month'] = df['date'].dt.to_period('M')
```

```python
# 1. Line Chart – Sales over Time
monthly_sales = df.groupby('month')['sales'].sum()
plt.figure()
monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("monthly_sales_line.png")
plt.show()
```

```python
# 2. Bar Chart – Category Comparison
# -----------------------------
category_sales = df.groupby('category')['sales'].sum()

plt.figure()
category_sales.plot(kind='bar')
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.savefig("category_sales_bar.png")
plt.show()
```

```python
# 3. Pie Chart – Category Share
# -----------------------------
plt.figure()
category_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Category Sales Share")
plt.ylabel("")
plt.tight_layout()
plt.savefig("category_sales_pie.png")
plt.show()
```
